```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = pd.read_excel(
    "/content/used_car_price_analysis_india_cleaned.xlsx"
)
```

**#USED CAR PRICE DISTRIBUTION**

```python
#Imported Matplotlib and Seaborn for data visualization.
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

#Customized the chart style, size, and appearance.
sns.set_theme(
    style="whitegrid",
    rc={
        "figure.dpi": 130,
        "axes.titlesize": 16,
        "axes.titleweight": "bold",
        "axes.labelsize": 11,
        "grid.alpha": 0.2,
        "axes.spines.top": False,
        "axes.spines.right": False
    }
)

#Create the figure and axes
fig, ax = plt.subplots(figsize=(11, 6))

sns.histplot(
    data=df_clean,
    x="listed_price_inr",
    bins=40,
    kde=True,
    color="#243B64",
    edgecolor="white",
    linewidth=0.5,
    ax=ax
)

#Display prices in lakhs
ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"₹{x / 100000:.0f}L")
)

ax.set_title(
    "Used Car Price Distribution",
    loc="left",
    pad=18
)
ax.set_xlabel("Listed Price")
ax.set_ylabel("Number of Listings")

ax.grid(axis="y", alpha=0.2)
ax.grid(axis="x", visible=False)

plt.tight_layout()
plt.show()
```

**#TOP 10 BRANDS BY AVERAGE LISTED PRICE**

```python
#Calculate the average listed price for each car brand
brand_price = (
    df_clean.groupby("brand")["listed_price_inr"]
    .mean()
    .nlargest(10)
    .sort_values()
)

#Create the figure and axes
fig, ax = plt.subplots(figsize=(11, 6))

brand_price.plot(
    kind="barh",
    ax=ax,
    color="#243B64",
    width=0.7
)

#Display prices in lakhs
ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"₹{x / 100000:.1f}L")
)

#Add chart title and axis labels
ax.set_title("Top 10 Brands by Average Listed Price", loc="left", pad=18)
ax.set_xlabel("Average Listed Price")
ax.set_ylabel("Car Brand")
ax.grid(axis="x", alpha=0.2)
ax.grid(axis="y", visible=False)

#Adjust layout and display chart
plt.tight_layout()
plt.show()
```
